```python
import pandas as pd
import matplotlib.pyplot as plt
import sys
import os
from pathlib import Path

def plot_histogram_from_csv(file_path, column_name, bins=10):
    """
    Reads a CSV file and plots a histogram for the specified numeric column.

    :param file_path: Path to the CSV file
    :param column_name: Name of the numeric column to plot
    :param bins: Number of bins for the histogram
    """
    # Validate file existence
    if not os.path.isfile(file_path):
        print(f"Error: File '{file_path}' not found.")
        sys.exit(1)

    try:
        # Read CSV into DataFrame
        df = pd.read_csv(file_path)
    except Exception as e:
        print(f"Error reading CSV file: {e}")
        sys.exit(1)

    # Validate column existence
    if column_name not in df.columns:
        print(f"Error: Column '{column_name}' not found in CSV.")
        print(f"Available columns: {list(df.columns)}")
        sys.exit(1)

    # Validate numeric data
    if not pd.api.types.is_numeric_dtype(df[column_name]):
        print(f"Error: Column '{column_name}' is not numeric.")
        sys.exit(1)

    # Plot histogram
    plt.figure(figsize=(8, 6))
    plt.hist(df[column_name].dropna(), bins=bins, edgecolor='black')
    plt.xlabel(column_name)
    plt.ylabel('Frequency')
    plt.title(f'Histogram of {column_name}')
    plt.grid(True, linestyle='--', alpha=0.6)
    plt.show()
```

```python
file_path = Path(r"C:\Users\pcadm\Downloads\SNN\SNN-Regression-Pencil-Balancer-True\Dataset\hough_cam1_cam1_hough.csv")
print(os.path.isfile(file_path))
column_name = "lin_m"
bins = 200

plot_histogram_from_csv(file_path, column_name, bins)

column_name = "lin_b"
plot_histogram_from_csv(file_path, column_name, bins)
```
